# Reproduce MLS data analysis

This notebook loads `mls_project_final.csv`, performs light cleaning and exploratory steps, and recreates two representative figures: a simple distribution plot (Figure_5) and a simplified Anemia dashboard.

Notes: this notebook is readability-first: code cells include comments and short explanations so others can reproduce and adapt the analysis.


```python
%matplotlib inline
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set(style="whitegrid")

# load the final dataset
df = pd.read_csv('../mls_project_final.csv', skipinitialspace=True)
df.columns = [c.strip() for c in df.columns]
df.head()
```

## Basic overview and cleaning
We show descriptive statistics and perform small cleaning steps (trim strings, normalize column types).


```python
# Basic summaries
print("shape:", df.shape)
print("columns:", df.columns.tolist())
display(df.describe(include='all'))

# cleaning examples: strip whitespace and normalize gender values
df['patient_name'] = df['patient_name'].astype(str).str.strip()
if 'age' in df.columns:
    df['age'] = pd.to_numeric(df['age'], errors='coerce')
if 'gender' in df.columns:
    df['gender'] = df['gender'].astype(str).str.strip().str.title()

df.head()
```

## Recreate Figure_5 (age distribution by blood group)
We'll produce a violin/box plot of age grouped by blood group as a representative figure.


```python
plt.figure(figsize=(10,6))
if 'blood_group' in df.columns:
    sns.boxplot(x='blood_group', y='age', data=df, palette='Set2')
    plt.title('Age distribution by blood group')
    plt.xlabel('Blood group')
    plt.ylabel('Age')
    plt.tight_layout()
    plt.savefig('../Figure_5_recreated.png', dpi=150)
    plt.show()
else:
    print('blood_group column not found in dataset')
```

## Simplified Anemia dashboard
We compute counts and anemia-classification-style summaries and render a small dashboard with key KPIs and a bar chart.


```python
# Example KPI: patient counts by blood group and gender
kpi_total = len(df)
kpi_by_blood = df['blood_group'].value_counts(dropna=False) if 'blood_group' in df.columns else None
kpi_by_gender = df['gender'].value_counts(dropna=False) if 'gender' in df.columns else None

print('Total patients:', kpi_total)
print('
Patients by blood group:
', kpi_by_blood)
print('
Patients by gender:
', kpi_by_gender)

# Bar chart for blood group counts
if kpi_by_blood is not None:
    plt.figure(figsize=(8,5))
    sns.barplot(x=kpi_by_blood.index, y=kpi_by_blood.values, palette='pastel')
    plt.title('Patient counts by blood group')
    plt.xlabel('Blood group')
    plt.ylabel('Count')
    plt.tight_layout()
    plt.savefig('../MLS_Anemia_Dashboard_recreated.png', dpi=150)
    plt.show()
```
